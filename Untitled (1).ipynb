```python
import numpy as np
import tensorflow.keras as tk
from keras import layers
import keras
import keras.datasets as ds
from datetime import datetime
from tensorflow.keras.utils import to_categorical
```

```python
logdir = 'logs/scalars' + datetime.now().strftime('%y%m%d - %H%M%S')
```

```python
%load_ext tensorboard
```

```python
tensorboard_callback = keras.callbacks.TensorBoard(log_dir = logdir)
```

```python
num_classes = 10
```

```python
inpunt_shape = (28,28,1)
```

```python
mnist = tk.datasets.mnist
```

```python
(x_train, y_train), (x_test, y_test) = mnist.load_data()
```

```python
x_train.shape, y_train.shape, x_test.shape, y_test.shape
```

```python
x_train = x_train.astype("float32")/255.
x_train = x_test.astype("float32")/255.
```

```python
x_train.shape, x_test.shape
```

```python
x_train = np.expand_dims(x_train, -1)
x_test = np.expand_dims(x_test, -1)
```

```python
x_train.shape, x_test.shape
```

```python
y_traian = keras.utils.to_categorical(y_train, num_classes)
y_test = keras.utils.to_categorical(y_train, num_classes)
```

```python
# Definir as variáveis necessárias
input_shape = (28, 28, 1)  # Exemplo para imagens em escala de cinza 28x28
num_classes = 10  # Exemplo para classificação de 10 categorias
```

```python
modelo = keras.Sequential(
    [
        keras.Input(shape=input_shape),  # Define a entrada do modelo
        layers.Conv2D(32, kernel_size=(3, 3), activation='relu'),  # Primeira camada convolucional
        layers.MaxPooling2D(pool_size=(2, 2)),  # Primeira camada de pooling
        layers.Conv2D(64, kernel_size=(3, 3), activation='relu'),  # Segunda camada convolucional
        layers.MaxPooling2D(pool_size=(2, 2)),  # Segunda camada de pooling
        layers.Flatten(),  # Achata as características extraídas pelas camadas anteriores
        layers.Dense(num_classes, activation='softmax')  # Camada densa final com ativação softmax
    ]
)
```

```python
batch_size = 100
epochs = 20
```

```python
modelo.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",  # Usando labels inteiros
    metrics=["accuracy"],
)
```

```python
training_history = modelo.fit(x_train,
                             y_train,
                             batch_size = batch_size,
                             epochs = epochs,
                             validation_split = 0.1,
                             callbacks = [tensorboard_callback]
                             )
```

```python
modelo.save('mnist_20_epochs.keras')
```
