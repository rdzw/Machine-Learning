```python
import cv2
import numpy as np

# Caminho para salvar os frames processados
output_dir = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frame_salvo_filtro\\'

# Função para aplicar o kernel
def aplicar_kernel(imagem, opcao):
    # Definindo 6 kernels diferentes
    kernels = {
        'a': np.array([[-1, -1, -1], [-1,  8, -1], [-1, -1, -1]]),  # Detecção de bordas
        'b': np.array([[ 0, -1,  0], [-1,  5, -1], [ 0, -1,  0]]),  # Filtro de nitidez
        'c': np.ones((5, 5), np.float32) / 25,                      # Filtro de média
        'd': np.array([[ 1,  4,  6,  4,  1], [ 4, 16, 24, 16,  4], [ 6, 24, 36, 24,  6], [ 4, 16, 24, 16,  4], [ 1,  4,  6,  4,  1]]) / 256,  # Filtro gaussiano
        'e': np.array([[ 0,  0,  0], [ 0,  1,  0], [ 0,  0,  0]]),  # Filtro de identidade
        'f': np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]),         # Filtro Sobel (detecção de bordas)
    }

    # Verificando se a opção é válida e aplicando o kernel
    if opcao in kernels:
        kernel = kernels[opcao]
        imagem_modificada = cv2.filter2D(imagem, -1, kernel)
        return imagem_modificada
    else:
        print("Opção inválida de kernel!")
        return imagem  # Retorna a imagem original se a opção for inválida

# Carrega a imagem
imagem = cv2.imread('C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frames\\1056.jpg')

# Processando a imagem com cada kernel e salvando
for opcao in ['a', 'b', 'c', 'd', 'e', 'f']:
    imagem_modificada = aplicar_kernel(imagem, opcao)
    
    # Exibe a imagem modificada
    cv2.imshow(f'Kernel {opcao}', imagem_modificada)
    
    # Salva a imagem modificada
    output_path = f'{output_dir}frame_1056_kernel_{opcao}.jpg'
    cv2.imwrite(output_path, imagem_modificada)
    print(f"Frame salvo: {output_path}")
    
    cv2.waitKey(0)

# Fecha as janelas
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Caminho para salvar os frames processados
output_dir = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frames_webcam_salvo_filtro\\'

# Função para aplicar o kernel
def aplicar_kernel(imagem, opcao):
    # Definindo 6 kernels diferentes
    kernels = {
        'a': np.array([[-1, -1, -1], [-1,  8, -1], [-1, -1, -1]]),  # Detecção de bordas
        'b': np.array([[ 0, -1,  0], [-1,  5, -1], [ 0, -1,  0]]),  # Filtro de nitidez
        'c': np.ones((5, 5), np.float32) / 25,                      # Filtro de média
        'd': np.array([[ 1,  4,  6,  4,  1], [ 4, 16, 24, 16,  4], [ 6, 24, 36, 24,  6], [ 4, 16, 24, 16,  4], [ 1,  4,  6,  4,  1]]) / 256,  # Filtro gaussiano
        'e': np.array([[ 0,  0,  0], [ 0,  1,  0], [ 0,  0,  0]]),  # Filtro de identidade
        'f': np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]),         # Filtro Sobel (detecção de bordas)
    }

    # Verificando se a opção é válida e aplicando o kernel
    if opcao in kernels:
        kernel = kernels[opcao]
        imagem_modificada = cv2.filter2D(imagem, -1, kernel)
        return imagem_modificada
    else:
        print("Opção inválida de kernel!")
        return imagem  # Retorna a imagem original se a opção for inválida

# Captura de vídeo da webcam
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("Erro ao acessar a webcam.")
    exit()

print("Pressione as teclas 'a', 'b', 'c', 'd', 'e', ou 'f' para aplicar os filtros.")
print("Pressione 'q' para sair.")

while True:
    ret, frame = cap.read()
    if not ret:
        print("Erro ao capturar o frame.")
        break

    cv2.imshow("Webcam Original", frame)

    key = cv2.waitKey(1) & 0xFF

    # Opções de filtros com teclas
    if chr(key) in ['a', 'b', 'c', 'd', 'e', 'f']:
        opcao = chr(key)
        frame_modificado = aplicar_kernel(frame, opcao)
        
        # Exibe o frame processado
        cv2.imshow(f"Filtro {opcao}", frame_modificado)
        
        # Salva o frame processado
        frame_path = f"{output_dir}frame_filtro_{opcao}.jpg"
        cv2.imwrite(frame_path, frame_modificado)
        print(f"Frame com filtro '{opcao}' salvo em: {frame_path}")

    # Saída ao pressionar 'q'
    if key == ord('q'):
        break

# Libera os recursos
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import os

# Diretório para salvar os frames processados
output_dir = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frames_redimensionados\\'
os.makedirs(output_dir, exist_ok=True)

# Constantes e configurações iniciais
REDIMENSIONAMENTO_INICIAL = 0.1  # Percentual inicial de redimensionamento (10%)
FATOR_MINIMO = 0.1  # 10%
FATOR_MAXIMO = 0.5  # 50%

# Mapeamento de teclas para tipos de interpolação
INTERPOLACOES = {
    'N': cv2.INTER_NEAREST,
    'L': cv2.INTER_LINEAR,
    'A': cv2.INTER_AREA,
    'C': cv2.INTER_CUBIC,
    'Z': cv2.INTER_LANCZOS4
}

# Função para redimensionar e salvar frames
def redimensionar_frames(fonte, tipo_interpolacao):
    fator_atual = REDIMENSIONAMENTO_INICIAL
    interpolacao_atual = INTERPOLACOES[tipo_interpolacao]
    
    # Abrindo a fonte (vídeo ou webcam)
    if isinstance(fonte, str):  # Caminho para vídeo
        cap = cv2.VideoCapture(fonte)
    else:  # Webcam
        cap = cv2.VideoCapture(0)
    
    if not cap.isOpened():
        print("Erro ao acessar a fonte de vídeo.")
        return
    
    print("Teclas interativas:")
    print("+ : Aumentar redimensionamento")
    print("- : Diminuir redimensionamento")
    print("N : INTER_NEAREST")
    print("L : INTER_LINEAR")
    print("A : INTER_AREA")
    print("C : INTER_CUBIC")
    print("Z : INTER_LANCZOS4")
    print("Q : Sair")
    
    frame_count = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            break
        
        # Redimensiona o frame
        novo_tamanho = (
            int(frame.shape[1] * fator_atual),
            int(frame.shape[0] * fator_atual)
        )
        frame_redimensionado = cv2.resize(frame, novo_tamanho, interpolation=interpolacao_atual)
        
        # Exibe o frame redimensionado
        cv2.imshow("Frame Redimensionado", frame_redimensionado)
        
        # Salva o frame redimensionado
        frame_path = os.path.join(output_dir, f'frame_{frame_count}_resized.jpg')
        cv2.imwrite(frame_path, frame_redimensionado)
        print(f"Frame salvo: {frame_path}")
        frame_count += 1
        
        # Captura a tecla pressionada
        key = cv2.waitKey(1) & 0xFF
        
        if key == ord('+'):  # Aumentar redimensionamento
            fator_atual = min(fator_atual + 0.1, FATOR_MAXIMO)
            print(f"Fator de redimensionamento atualizado para: {fator_atual:.1%}")
        elif key == ord('-'):  # Diminuir redimensionamento
            fator_atual = max(fator_atual - 0.1, FATOR_MINIMO)
            print(f"Fator de redimensionamento atualizado para: {fator_atual:.1%}")
        elif chr(key).upper() in INTERPOLACOES:  # Alterar interpolação
            interpolacao_atual = INTERPOLACOES[chr(key).upper()]
            print(f"Tipo de interpolação alterado para: {chr(key).upper()}")
        elif key == ord('q'):  # Sair
            break
    
    # Libera os recursos
    cap.release()
    cv2.destroyAllWindows()

# Exemplo de uso:
# Para webcam: passe `None` como primeiro parâmetro.
# Para vídeo: passe o caminho do vídeo como primeiro parâmetro.
# Segundo parâmetro deve ser uma das teclas iniciais ('N', 'L', 'A', 'C', 'Z') para o tipo de interpolação inicial.

redimensionar_frames(None, 'L')  # Webcam com interpolação inicial INTER_LINEAR
```

```python
import cv2
import os

# Diretório para salvar as fotos
output_dir = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\fotos_produto\\'
os.makedirs(output_dir, exist_ok=True)

# Inicializar a captura de vídeo
cap = cv2.VideoCapture(0)  # 0 para usar a webcam padrão, ou substitua pelo caminho de um arquivo de vídeo

if not cap.isOpened():
    print("Erro ao acessar a câmera.")
    exit()

# Mensagem de instruções
print("Pressione 's' para tirar uma foto.")
print("Pressione 'q' para sair.")

while True:
    # Captura um frame da câmera
    ret, frame = cap.read()
    if not ret:
        print("Falha ao capturar o frame.")
        break

    # Exibe o frame capturado em tempo real
    cv2.imshow("Captura de Foto", frame)

    # Captura a tecla pressionada
    key = cv2.waitKey(1) & 0xFF
    
    if key == ord('s'):  # Quando pressionar 's', salva a foto
        foto_path = os.path.join(output_dir, 'foto_produto.jpg')
        cv2.imwrite(foto_path, frame)
        print(f"Foto salva em: {foto_path}")
        
    elif key == ord('q'):  # Quando pressionar 'q', sai do loop
        break

# Libera a captura e fecha as janelas
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import os

# Diretório para salvar as fotos
output_dir = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\fotos_produto\\'
os.makedirs(output_dir, exist_ok=True)

# Inicializar a captura de vídeo
cap = cv2.VideoCapture(0)  # 0 para usar a webcam padrão, ou substitua pelo caminho de um arquivo de vídeo

if not cap.isOpened():
    print("Erro ao acessar a câmera.")
    exit()

# Mensagem de instruções
print("Pressione 's' para tirar uma foto.")
print("Pressione 'q' para sair.")

# Primeiro, capture a foto do produto com todas as peças
while True:
    # Captura um frame da câmera
    ret, frame = cap.read()
    if not ret:
        print("Falha ao capturar o frame.")
        break

    # Exibe o frame capturado em tempo real
    cv2.imshow("Captura de Foto - Produto Completo", frame)

    # Captura a tecla pressionada
    key = cv2.waitKey(1) & 0xFF
    
    if key == ord('s'):  # Quando pressionar 's', salva a foto
        foto_path = os.path.join(output_dir, 'foto_produto_completo.jpg')
        cv2.imwrite(foto_path, frame)
        print(f"Foto salva em: {foto_path}")
        break
        
    elif key == ord('q'):  # Quando pressionar 'q', sai do loop
        break



print("Remova uma ou duas peças do produto, garantindo que o restante não seja movido.")

while True:
    # Captura um frame da câmera após a remoção das peças
    ret, frame = cap.read()
    if not ret:
        print("Falha ao capturar o frame.")
        break

    # Exibe o frame capturado em tempo real
    cv2.imshow("Captura de Foto - Produto Parcial", frame)

    # Captura a tecla pressionada
    key = cv2.waitKey(1) & 0xFF
    
    if key == ord('s'):  # Quando pressionar 's', salva a foto
        foto_path = os.path.join(output_dir, 'foto_produto_parcial.jpg')
        cv2.imwrite(foto_path, frame)
        print(f"Foto salva em: {foto_path}")
        break
        
    elif key == ord('q'):  # Quando pressionar 'q', sai do loop
        break

# Libera os recursos
cap.release()
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Função para plotar 3 imagens lado a lado
def plot_3img(im1, tit1, im2, tit2, im3, tit3, tam):
    plt.figure(figsize=tam)
    plt.subplot(1, 3, 1), plt.imshow(im1), plt.title(tit1)
    plt.xticks([]), plt.yticks([])  # Remove ticks
    plt.subplot(1, 3, 2), plt.imshow(im2), plt.title(tit2)
    plt.xticks([]), plt.yticks([])  # Remove ticks
    plt.subplot(1, 3, 3), plt.imshow(im3), plt.title(tit3)
    plt.xticks([]), plt.yticks([])  # Remove ticks
    plt.show()

# Caminho completo para as imagens
foto_completa_path = 'C:/Users/rodrigo/Desktop/Machine Learning Aplicada/T1/fotos_produto/foto_produto_completo.jpg'
foto_parcial_path = 'C:/Users/rodrigo/Desktop/Machine Learning Aplicada/T1/fotos_produto/foto_produto_parcial.jpg'

# Carregar as duas fotos (completa e parcial)
foto_completa = cv2.imread(foto_completa_path)
foto_parcial = cv2.imread(foto_parcial_path)

# Verificar se as imagens foram carregadas corretamente
if foto_completa is None:
    print(f"Erro ao carregar a foto completa do caminho: {foto_completa_path}")
    exit()

if foto_parcial is None:
    print(f"Erro ao carregar a foto parcial do caminho: {foto_parcial_path}")
    exit()

# Converter para escala de cinza para facilitar a comparação
foto_completa_gray = cv2.cvtColor(foto_completa, cv2.COLOR_BGR2GRAY)
foto_parcial_gray = cv2.cvtColor(foto_parcial, cv2.COLOR_BGR2GRAY)

# Subtração de imagens
diferenca = cv2.absdiff(foto_completa_gray, foto_parcial_gray)

# Aplicar limiarização para destacar as diferenças
_, diferenca_binaria = cv2.threshold(diferenca, 50, 255, cv2.THRESH_BINARY)

# Encontrar contornos nas diferenças
contornos, _ = cv2.findContours(diferenca_binaria, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

# Copiar a imagem completa para desenhar os retângulos
foto_completa_com_quadrados = foto_completa.copy()

# Refinar os contornos e desenhar retângulos com precisão
for contorno in contornos:
    # Ignorar áreas muito pequenas (ruídos)
    area = cv2.contourArea(contorno)
    if area > 100:  # Ajuste o valor conforme necessário
        # Encaixar um retângulo ao contorno
        x, y, w, h = cv2.boundingRect(contorno)
        
        # Ajustar bordas para melhor precisão
        margem = 5  # Adiciona uma margem extra
        x = max(0, x - margem)
        y = max(0, y - margem)
        w += 2 * margem
        h += 2 * margem

        # Desenhar retângulo refinado diretamente na imagem original
        cv2.rectangle(foto_completa_com_quadrados, (x, y), (x + w, y + h), (0, 0, 255), 2)

# Usar a função plot_3img para mostrar as imagens
plot_3img(
    cv2.cvtColor(foto_completa, cv2.COLOR_BGR2RGB), 'Foto Completa',
    cv2.cvtColor(foto_parcial, cv2.COLOR_BGR2RGB), 'Foto Parcial',
    cv2.cvtColor(foto_completa_com_quadrados, cv2.COLOR_BGR2RGB), 'Diferenças (Quadrados Precisos)',
    (12, 10)
)
```
