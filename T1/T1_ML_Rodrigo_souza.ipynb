```python
#importando o opencv
import cv2
```

```python
# Caminho para o vídeo
video_path = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\video\\teste.mp4'

# Criação do objeto de captura de vídeo
video = cv2.VideoCapture(video_path)

# Verifica se o vídeo foi carregado corretamente
if not video.isOpened():
    print("Erro: Não foi possível abrir o vídeo!")
    exit()

# Processa o vídeo frame por frame
while True:
    check, img = video.read()
    
    # Verifica se há mais frames para processar
    if not check:
        print("Fim do vídeo ou erro na leitura!")
        break

    # Redimensiona o frame
    imgRedin = cv2.resize(img, (640, 420))
    
    # Mostra o frame redimensionado
    cv2.imshow('video', imgRedin)
    
    # Aguarda 10ms e verifica se a tecla 'q' foi pressionada para sair
    if cv2.waitKey(10) & 0xFF == ord('q'):
        break

# Libera o vídeo e fecha as janelas
video.release()
cv2.destroyAllWindows()
```

```python
# Gerendo frames
# Caminho do vídeo
video_path = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\video\\teste.mp4'
# Caminho onde os frames serão salvos
frames_path = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frames\\'

# Abre o vídeo
cap = cv2.VideoCapture(video_path)

# Verifica se o vídeo foi aberto corretamente
if not cap.isOpened():
    print("Erro: Não foi possível abrir o vídeo!")
    exit()

cont = 1

# Lê e salva os frames enquanto o vídeo tiver frames
while True:
    ret, frame = cap.read()
    
    if not ret:
        break  # Sai do loop se não houver mais frames

    # Salva o frame com o nome sequencial
    cv2.imwrite(frames_path + str(cont) + '.jpg', frame)
    print(f'Frame {cont} salvo.')
    cont += 1

# Libera o vídeo e fecha as janelas
cap.release()
cv2.destroyAllWindows()

print('Fim da geração dos frames.')
```

```python
import numpy as np
```

```python
# Parte A: Aplicação de filtros
# Função para aplicar o kernel
def aplicar_kernel(imagem, opcao):
    # Definindo 6 kernels diferentes
    kernels = {
        'a': np.array([[-1, -1, -1], [-1,  8, -1], [-1, -1, -1]]),  # Detecção de bordas
        'b': np.array([[ 0, -1,  0], [-1,  5, -1], [ 0, -1,  0]]),  # Filtro de nitidez
        'c': np.ones((5, 5), np.float32) / 25,                      # Filtro de média
        'd': np.array([[ 1,  4,  6,  4,  1], [ 4, 16, 24, 16,  4], [ 6, 24, 36, 24,  6], [ 4, 16, 24, 16,  4], [ 1,  4,  6,  4,  1]]) / 256,  # Filtro gaussiano
        'e': np.array([[ 0,  0,  0], [ 0,  1,  0], [ 0,  0,  0]]),  # Filtro de identidade
        'f': np.array([[-1, 0, 1], [-2, 0, 2], [-1, 0, 1]]),  # Filtro Sobel (detecção de bordas)
    }

    # Verificando se a opção é válida e aplicando o kernel
    if opcao in kernels:
        kernel = kernels[opcao]
        imagem_modificada = cv2.filter2D(imagem, -1, kernel)
        return imagem_modificada
    else:
        print("Opção inválida de kernel!")
        return imagem  # Retorna a imagem original se a opção for inválida

# Carrega a imagem
imagem = cv2.imread('C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\frames\\1056.jpg')

# Testando a função com diferentes kernels
for opcao in ['a', 'b', 'c', 'd', 'e', 'f']:
    imagem_modificada = aplicar_kernel(imagem, opcao)
    
    # Exibe a imagem modificada
    cv2.imshow(f'Kernel {opcao}', imagem_modificada)
    cv2.waitKey(0)

# Fecha as janelas
cv2.destroyAllWindows()
```
