```python
#importando o opencv
import cv2
```

```python
# Caminho para o vídeo
video_path = 'C:\\Users\\rodrigo\\Desktop\\Machine Learning Aplicada\\T1\\video\\teste.mp4'

# Criação do objeto de captura de vídeo
video = cv2.VideoCapture(video_path)

# Verifica se o vídeo foi carregado corretamente
if not video.isOpened():
    print("Erro: Não foi possível abrir o vídeo!")
    exit()

# Processa o vídeo frame por frame
while True:
    check, img = video.read()
    
    # Verifica se há mais frames para processar
    if not check:
        print("Fim do vídeo ou erro na leitura!")
        break

    # Redimensiona o frame
    imgRedin = cv2.resize(img, (640, 420))
    
    # Mostra o frame redimensionado
    cv2.imshow('video', imgRedin)
    
    # Aguarda 10ms e verifica se a tecla 'q' foi pressionada para sair
    if cv2.waitKey(10) & 0xFF == ord('q'):
        break

# Libera o vídeo e fecha as janelas
video.release()
cv2.destroyAllWindows()
```
