# Importando bibliotecas adicionais para treinamento

```python
import os
import pandas as pd
import tensorflow as tf
from tensorflow.keras.applications.resnet50 import ResNet50, preprocess_input
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.layers import Dense, GlobalAveragePooling2D
from sklearn.model_selection import train_test_split
from tensorflow.keras.preprocessing import image
import numpy as np
import matplotlib.pyplot as plt
```

# Número de classes do problema: gatos e cachorros

```python
num_classes = 2
```

# Definindo a semente para reprodutibilidade

```python
tf.random.set_seed(0)
```

# Carregando o modelo base (ResNet50) sem a camada final (include_top=False)

```python
modelo_base = ResNet50(weights = 'imagenet', include_top = False, input_shape = (224, 224, 3))
```

# Congelando os pesos do modelo base para evitar ajuste durante o treinamento

```python
for layer in modelo_base.layers:
    layer.trainable = False
```

# Criando o modelo final adicionando camadas ao modelo base

```python
modelo = tf.keras.Sequential(
    [modelo_base,
     GlobalAveragePooling2D(),
     Dense(1024, activation = 'relu'),
     Dense(num_classes, activation = 'softmax')
    ])
```

# Caminho para o diretório das imagens

```python
path = '/home/turma02/Imagens/dc_small/'
```

# Função para criar um DataFrame com arquivos e rótulos

```python
def cria_dataframe(caminho):
    arquivos = []
    rotulos = []
    for arquivo in os.listdir(caminho):
        if arquivo.startswith('cat'):  # Corrigido para acessar o nome do arquivo
            rotulos.append('cat')
        elif arquivo.startswith('dog'):  # Corrigido para acessar o nome do arquivo
            rotulos.append('dog')
        arquivos.append(arquivo)
    return pd.DataFrame({'arquivo': arquivos, 'rotulo': rotulos}) 
```

# Criando o DataFrame a partir do diretório de treinamento

```python
df_dados = cria_dataframe(os.path.join(path, 'train'))
```

# Exibindo informações do DataFrame

```python
df_dados.info()
```

# Dividindo o DataFrame em treinamento e validação

```python
train_df, validation_df = train_test_split(df_dados, test_size = 0.2,
                                           stratify = df_dados['rotulo'], random_state = 0)
```

# Criando o DataFrame para o conjunto de teste

```python
test_df = cria_dataframe(os.path.join(path, 'test'))
```

# Configurando os geradores de dados para pré-processar as imagens

```python
train_datagen = ImageDataGenerator(preprocessing_function = preprocess_input)
validation_datagen = ImageDataGenerator(preprocessing_function = preprocess_input)
test_datagen = ImageDataGenerator(preprocessing_function = preprocess_input)
```

# Criando os geradores de dados para treinamento, validação e teste

```python
train_generator = train_datagen.flow_from_dataframe(
    train_df,
    directory = os.path.join(path, 'train'),
    x_col = 'arquivo',
    y_col = 'rotulo',
    target_size = (224, 224),
    batch_size = 64,
    class_mode = 'categorical'
)
```

```python
validation_generator = validation_datagen.flow_from_dataframe(
    test_df,
    directory = os.path.join(path, 'test'),
    x_col = 'arquivo',
    y_col = 'rotulo',
    target_size = (224, 224),
    batch_size = 64,
    class_mode = 'categorical'
)
```

```python
test_generator = test_datagen.flow_from_dataframe(
    test_df,
    directory = os.path.join(path, 'test'),
    x_col = 'arquivo',
    y_col = 'rotulo',
    target_size = (224, 224),
    batch_size = 64,
    class_mode = 'categorical'
)
```

# Compilando o modelo com otimizador Adam, função de perda e métrica de acurácia

```python
modelo.compile(optimizer = tf.keras.optimizers.Adam(learning_rate=0.001),
                                                    loss='categorical_crossentropy',
                                                    metrics=['accuracy'])
```

# Treinando o modelo

```python
modelo.fit(
    train_generator,
    steps_per_epoch = train_generator.samples // train_generator.batch_size,
    validation_data = validation_generator,
    validation_steps = validation_generator.samples // validation_generator.batch_size,
    epochs = 10
)
```

# Avaliando o modelo no conjunto de teste

```python
test_loss, test_accuracy = modelo.evaluate(test_generator, steps = test_generator.samples // test_generator.batch_size)
print(f'Acuracia no Teste: {test_accuracy}')
```

# Salvando o modelo treinado

```python
modelo.save('resnet_10_epoch_dogs_cats.keras')
```

# Função para realizar a predição em uma imagem e retornar o rótulo e a imagem processada

```python
def predicao_img(imagem, modelo):
    img = image.load_img(imagem, target_size = (224, 224))
    img_array = image.img_to_array(img)
    img_array = np.expand_dims(img_array, axis = 0)
    img_array = preprocess_input(img_array)
    predicoes = modelo.predict(img_array)
    classes = {0: 'cat', 1: 'dog'}
    classes_predita = classes[np.argmax(predicoes[0])]
    return classes_predita, img
```

# Realizando uma predição em uma imagem de amostra

# Exibindo a imagem de amostra com a classe predita

```python
amostra = path + 'img9.jpeg'
classe_predita, img_resposta = predicao_img(amostra, modelo)
plt.figure(figsize = (3, 3))
plt.imshow(img_resposta)
plt.title(f'Classe: {classe_predita}')
plt.show()
```
