```python
from sklearn import linear_model
import seaborn as sns
```

```python
df_selecao = df_co2[['ENGINESIZE', 'CO2EMISSIONS']]
```

```python
rl = sns.scatterplot(x = 'ENGINESIZE', y = 'CO2EMISSIONS', data = df_selecao,
                     hue = 'ENGINESIZE')
```

```python
len(df_selecao)
```

```python
treino1 = df_selecao[:(int(len(df_selecao)*0.7))]
teste1 = df_selecao[:(int(len(df_selecao)*0.7)):]
```

```python
treino1.shape, teste1.shape
```

```python
modelo_rl1 = linear_model.LinearRegression()
```

```python
import numpy as np
```

```python
treino_X = np.array(treino1[['ENGINESIZE']])
treino_y = np.array(treino1[['CO2EMISSIONS']])
```

```python
modelo_rl1.fit(treino_X,treino_y)
```

```python
modelo_rl1.coef_, modelo_rl1.intercept_
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(treino1['ENGINESIZE'], treino1['CO2EMISSIONS'], color ='green')
plt.plot(treino_X, modelo_rl1.coef_*treino_X + modelo_rl1.intercept_, '-r')
plt.xlabel('Engine Size')
plt.ylabel('Emissão CO2')
plt.show()

```

```python
def predicao_co2(dado, intercept, angulo):
    pred = dado * angulo + intercept
    return pred
```

```python
potencia = 1.8
emissao_estimada =  predicao_co2(potencia,
                                 modelo_rl1.intercept_[0],
                                 modelo_rl1.coef_[0][0])
print('A emissão de Co2 para o veículo de potência', potencia, 'é: ', emissao_estimada)
```

```python
import numpy as np
from sklearn.datasets import load_digits
from matplotlib import pyplot as plt
from sklearn.neighbors import KNeighborsClassifier as KNN
from sklearn.model_selection import train_test_split
from sklearn import metrics
```

```python
digitos = load_digits()
```

```python
len(digitos.feature_names)
```

```python
digitos.target
```

```python
digitos.target_names
```

```python
digitos.images
```

```python
fig = plt.figure(figsize =(8,6))
fig.subplots_adjust(left =0, right =1, bottom =0, top=1,
                    hspace=0.05, wspace= 0.05)
for i in range(25):
    ax = fig.add_subplot(5, 5, i +1, xticks =[], yticks =[])
    ax.imshow(digitos.images[i], cmap = plt.cm.gray, interpolation= 'lanczos')
```

```python
X_dig = digitos.data
y_dig = digitos.target
```

```python
X_dig.shape, y_dig.shape
```

```python
X_train_dig, X_test_dig, y_train_dig, y_test_dig = train_test_split(
    X_dig, y_dig, test_size= 0.3, random_state=0)
```

```python
X_train_dig.shape, X_test_dig.shape, y_train_dig.shape, y_test_dig.shape
```

```python
modelo_knn = KNN(n_neighbors =5)
```

```python
modelo_knn.fit(X_train_dig, y_train_dig)
```

```python
predis_dig = modelo_knn.predict(X_test_dig)
```

```python
acuracia_knn = modelo_knn.score(X_test_dig, y_test_dig)
print('Acurácia do KNN:' , acuracia_knn)
```

```python
matriz_conf_KNN = metrics.confusion_matrix(y_test_dig, predis_dig)
print('Matriz de confusão:')
print(matriz_conf_KNN)
```

```python
rela_class_KNN = metrics.classification_report(y_test_dig, predis_dig)
print('Relatorio de classificação:')
print(rela_class_KNN)
```
