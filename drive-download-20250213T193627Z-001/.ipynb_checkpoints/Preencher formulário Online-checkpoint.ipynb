```python
import pandas as pd
from IPython.display import display
import pandas as pd

# Carrega o arquivo Excel
f = pd.read_excel("2.Mechanical Sampling_v2 (P).xlsx")

# Preenche células mescladas com os valores superiores
f.ffill(inplace=True)

# Definição correta do nome da coluna a ser filtrada
coluna_filtro = "padrão número (Rev.)Data de revisão"

# Aplicar filtro corretamente com regex
filtro = f[f[coluna_filtro].astype(str).str.contains(r"LG\(53\)-G7-2050|LG\(51\)-G7-2050", na=False, regex=True)]

# Seleciona todas as colunas a partir da 5ª
dados_filtrados = filtro.iloc[:, 5:]

# Remove a coluna diretamente
dados_filtrados.drop(dados_filtrados.columns[2], axis=1, inplace=True)
dados_filtrados.drop(dados_filtrados.columns[6], axis=1, inplace=True)
dados_filtrados.drop(dados_filtrados.columns[8], axis=1, inplace=True)

dados_filtrados = dados_filtrados.iloc[:, :7]  # Mantém apenas as 9 primeiras colunas

# Exibir os dados formatados
display(dados_filtrados)
```

```python
for linha in dados_filtrados.itertuples(index=False, name=None):
    print(linha[2])  # Exibe a tupla com os valores da linha
```
